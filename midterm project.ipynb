## Section 3: The Model (25 pts)

Fit a multivariate OLS model and report it in a formatted regression table (coefficients, standard errors, N, R-squared), not a `summary()` dump. Interpret at least two coefficients in full: magnitude, units, direction, precision. For every control, say whether it blocks a confounding path or adds precision, and what would make it a bad control (Chapter 10). OLS only: no difference-in-differences estimate.

```python
m1 = smf.ols("life_expectancy ~ log_health", data=sample).fit(cov_type="HC1")
m2 = smf.ols("life_expectancy ~ log_health + urban_population_pct + log_gdp",
             data=sample).fit(cov_type="HC1")

regression_table = pd.DataFrame({
    "Coefficient": m2.params,
    "Robust SE": m2.bse,
    "p-value": m2.pvalues,
    "95% CI low": m2.conf_int()[0],
    "95% CI high": m2.conf_int()[1]
})
print("Main model: life expectancy in years")
display(regression_table.round(3))
print("N:", int(m2.nobs))
print("R-squared:", round(m2.rsquared, 3))
print("Health coefficient without controls:", round(m1.params["log_health"], 3))
```

```python
health_doubling = m2.params["log_health"] * np.log(2)
health_ci = m2.conf_int().loc["log_health"] * np.log(2)

urban_ten = m2.params["urban_population_pct"] * 10
urban_ci = m2.conf_int().loc["urban_population_pct"] * 10

print("Years per doubling of health spending:", round(health_doubling, 2))
print("95% CI:", round(health_ci.iloc[0], 2), "to", round(health_ci.iloc[1], 2))
print("Years per 10 percentage points more urban:", round(urban_ten, 2))
print("95% CI:", round(urban_ci.iloc[0], 2), "to", round(urban_ci.iloc[1], 2))
```

*Your words here.*

Among economies with similar GDP per person and urbanization, twice the health spending per person is associated with 1.35 more years of life expectancy. The 95% confidence interval is 0.12 to 2.58 years, so the estimate is positive but fairly imprecise. After accounting for health spending and GDP, a 10-percentage-point higher urban share is associated with 0.02 fewer years of life expectancy. Its 95% interval ranges from 0.45 fewer to 0.41 more years, leaving the direction of the urbanization relationship uncertain. The regression table uses HC1 robust standard errors. These figures describe differences across economies; they do not tell us what would happen if a country changed its health budget.

I control for GDP per person because national income can affect both health spending and survival. GDP could be a bad control for a total causal effect if earlier health spending also affected income. Urbanization is my second predictor, but if health spending affects urbanization, controlling for it could remove part of the spending effect.

## Section 4: Diagnostics (20 pts)

Residuals against fitted values and against your main predictor; a constant-variance check, with robust standard errors if it fails; variance inflation factors, interpreted; at least one influential-observation check.

**Start only after Topic 14.** Write Sections 1 to 3 first; they do not depend on it.

```python
plain = smf.ols("life_expectancy ~ log_health + urban_population_pct + log_gdp",
                data=sample).fit()

fig, ax = plt.subplots(1, 2, figsize=(10, 3.4))
ax[0].scatter(plain.fittedvalues, plain.resid)
ax[0].axhline(0, color="grey")
ax[0].set(xlabel="Fitted life expectancy (years)", ylabel="Residual (years)",
          title="Residuals versus fitted values")
ax[1].scatter(sample["log_health"], plain.resid)
ax[1].axhline(0, color="grey")
ax[1].set(xlabel="Log of health spending", ylabel="Residual (years)",
          title="Residuals versus health spending")
plt.tight_layout()
plt.show()
```

```python
bp_p = het_breuschpagan(plain.resid, plain.model.exog)[1]
print("Breusch-Pagan p-value:", round(bp_p, 4))
print("The main regression table uses HC1 robust standard errors.")

X = plain.model.exog
names = plain.model.exog_names
vif_values = {}
for i in range(1, len(names)):
    vif_values[names[i]] = variance_inflation_factor(X, i)
    print("VIF for", names[i], ":", round(vif_values[names[i]], 2))
```

```python
cooks = pd.Series(plain.get_influence().cooks_distance[0],
                  index=sample["country_name"])
cutoff = 4 / len(sample)
print("Cook's distance screening line (4/N):", round(cutoff, 3))
print("Largest Cook's distances:")
print(cooks.nlargest(5).round(3))

largest = cooks.idxmax()
without_largest = sample[sample["country_name"] != largest]
m_without = smf.ols("life_expectancy ~ log_health + urban_population_pct + log_gdp",
                    data=without_largest).fit(cov_type="HC1")
print("Most influential economy:", largest)
print("Health coefficient, all economies:", round(m2.params["log_health"], 3))
print("Health coefficient, without it:", round(m_without.params["log_health"], 3))
```

The residual plots show one large negative residual, and the spread looks uneven. The Breusch-Pagan p-value is 0.0073, which is evidence against constant residual variance. I use HC1 robust standard errors in the main regression table, as in the teacher’s example. The VIFs are 9.09 for log health spending, 8.97 for log GDP, and 1.99 for urban share. Health spending and GDP overlap strongly, making it harder to estimate their separate relationships with life expectancy precisely.

The Central African Republic has the largest Cook’s distance at 0.481, above the 4/N screening line of 0.022. Its life expectancy is 31.53 years in the source file. Leaving it out changes the health spending coefficient from 1.950 to 2.311. I keep it in the main model and report the comparison because its influence alone is not evidence that the source value is wrong.
